```python
# calculate trip duration and drop tpep_dropoff_datetime and related features (only known after the trip)
df["duration"] = (df["tpep_dropoff_datetime"] - df["tpep_pickup_datetime"]).dt.total_seconds() / 60

leaky_cols = [
    "tpep_dropoff_datetime",
    "fare_amount",
    "extra",
    "mta_tax",
    "tip_amount",
    "tolls_amount",
    "improvement_surcharge",
    "total_amount",
    "congestion_surcharge",
    "Airport_fee",
    "cbd_congestion_fee",
    "payment_type",
    "store_and_fwd_flag",
    "trip_distance",  # remove this line if you'll have a distance estimate at pickup
]
df = df.drop(columns=leaky_cols)
```

```python
# filter outliers with fixed cutoff
print(df.shape)
filtered_df = df[df["duration"].between(5, 120)]
print(filtered_df.shape)
```

```python
from sklearn.model_selection import train_test_split  # type: ignore

X = filtered_df.drop(columns=["duration"])
y = filtered_df["duration"]

X_train, X_test, y_train, y_test = train_test_split(X, y, random_state=25) 

print(X_train.shape)
print(X_test.shape)
print(y_train.shape)
print(y_test.shape)
```

```python
X_train.nunique()
```

```python
from sklearn.compose import ColumnTransformer
from sklearn.ensemble import RandomForestRegressor
from sklearn.impute import SimpleImputer
from sklearn.linear_model import LinearRegression
from sklearn.metrics import root_mean_squared_error
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import FunctionTransformer, OneHotEncoder, TargetEncoder
```

```python
cat_cols = ["VendorID", "RatecodeID", "PULocationID", "DOLocationID"]
num_cols = ["passenger_count"]

categorical_pipeline = Pipeline([
    ("imputer", SimpleImputer(strategy="most_frequent")),
    ("encoder", OneHotEncoder(handle_unknown="ignore"))
])

numerical_pipeline = Pipeline([
    ("imputer", SimpleImputer(strategy="most_frequent"))
])

preprocess = ColumnTransformer([
    ("categorical", categorical_pipeline, cat_cols),
    ("numerical", numerical_pipeline, num_cols)
])

base_model = Pipeline([
    ("preprocess", preprocess),
    ("regressor", LinearRegression())
])

base_model.fit(X_train, y_train)

predictions = base_model.predict(X_test)

rmse = root_mean_squared_error(y_test, predictions)

print("base model RMSE: ", rmse)
```

```python
# FEATURE ENGINEERING (add time features and route, drop tpep_pickup_datetime)
def add_features(X):
    X = X.copy()
    pickup = X["tpep_pickup_datetime"]
    X["hour"] = pickup.dt.hour
    X["day_of_week"] = pickup.dt.dayofweek
    X["is_weekend"] = (X["day_of_week"] >= 5).astype(int)
    X["route"] = X["PULocationID"].astype(str) + "_" + X["DOLocationID"].astype(str)
    return X.drop(columns=["tpep_pickup_datetime"])
```

```python
target_cols = ["PULocationID", "DOLocationID", "route"]
onehot_cols = ["VendorID", "RatecodeID"]
num_cols = ["passenger_count", "hour", "day_of_week", "is_weekend"]

target_pipeline = Pipeline([
    ("encoder", TargetEncoder())
])

onehot_pipeline = Pipeline([
    ("imputer", SimpleImputer(strategy="most_frequent")),
    ("onehot", OneHotEncoder(handle_unknown="ignore"))
])

num_pipeline = Pipeline([
    ("imputer", SimpleImputer(strategy="most_frequent"))
])

preprocess = ColumnTransformer([
    ("target", target_pipeline, target_cols),
    ("onehot", onehot_pipeline, onehot_cols),
    ("numerical", num_pipeline, num_cols)
])

random_forest = Pipeline([
    ("features", FunctionTransformer(add_features)),
    ("preprocess", preprocess),
    ("random_forest", RandomForestRegressor(
        n_estimators=100,
        max_depth=20,
        min_samples_leaf=20,
        max_samples=0.1,   # each tree sees 10% of rows -> much faster
        n_jobs=-1,
        random_state=25,
    ))
])

random_forest.fit(X_train, y_train)
random_forest_predictions = random_forest.predict(X_test)
random_forest_rmse = root_mean_squared_error(y_test, random_forest_predictions)

print("random_forest_rmse: ", random_forest_rmse)
```

```python
from sklearn.base import clone
from sklearn.ensemble import HistGradientBoostingRegressor

# makes a fresh copy of the preprocessing. Reusing preprocess directly would refit the object inside your random forest pipeline.
hgb_preprocess = clone(preprocess).set_params(onehot__onehot__sparse_output=False)

hgb_model = Pipeline([
    ("features", FunctionTransformer(add_features)),
    ("preprocess", hgb_preprocess),
    ("hgb", HistGradientBoostingRegressor(
        max_iter=300,
        learning_rate=0.1,
        max_leaf_nodes=63,
        early_stopping=True,
        random_state=25,
    ))
])

hgb_model.fit(X_train, y_train)
hgb_predictions = hgb_model.predict(X_test)
hgb_rmse = root_mean_squared_error(y_test, hgb_predictions)

print("hgb_rmse: ", hgb_rmse)
```

```python
results = pd.DataFrame({
    "model": ["Linear regression (baseline)", "Random forest", "HistGradientBoosting"],
    "rmse": [rmse, random_forest_rmse, hgb_rmse],
})
results["improvement_vs_baseline"] = (1 - results["rmse"] / rmse).map("{:.1%}".format)
results = results.sort_values("rmse").round(2)

print(results)
```

```python
from sklearn.model_selection import RandomizedSearchCV

param_dist = {
    "hgb__learning_rate": [0.03, 0.05, 0.1, 0.2],
    "hgb__max_leaf_nodes": [31, 63, 127, 255],
    "hgb__min_samples_leaf": [20, 50, 100, 200],
    "hgb__l2_regularization": [0, 0.1, 1.0],
}

# Search on a 500k-row sample to keep it fast
X_sample = X_train.sample(500_000, random_state=25)
y_sample = y_train.loc[X_sample.index]

search = RandomizedSearchCV(
    hgb_model,
    param_dist,
    n_iter=15,
    cv=3,
    scoring="neg_root_mean_squared_error",
    n_jobs=-1,
    random_state=25,
    verbose=1,
)
search.fit(X_sample, y_sample)

print("Best params:", search.best_params_)
print("CV RMSE:   ", -search.best_score_)
```

```python
best_hgb = clone(hgb_model).set_params(**search.best_params_)
best_hgb.fit(X_train, y_train)
tuned_rmse = root_mean_squared_error(y_test, best_hgb.predict(X_test))
print("Tuned HGB RMSE:", tuned_rmse)
```

```python
results = pd.DataFrame({
    "model": ["Linear regression (baseline)", "Random forest", "HistGradientBoosting", "Tuned HistGradientBoosting"],
    "rmse": [rmse, random_forest_rmse, hgb_rmse, tuned_rmse],
})
results["improvement_vs_baseline"] = (1 - results["rmse"] / rmse).map("{:.1%}".format)
results = results.sort_values("rmse").round(2)

print(results)
```
